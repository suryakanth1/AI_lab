```python
import heapq

# Misplaced tiles heuristic
def misplaced(state, goal):
    return sum(1 for i, tile in enumerate(state) if tile != 0 and tile != goal[i])

# Generate neighbors by sliding blank
def neighbors(state):
    moves = []
    i = state.index(0)  # blank position
    r, c = divmod(i, 3)
    for dr, dc in [(-1,0),(1,0),(0,-1),(0,1)]:
        nr, nc = r + dr, c + dc
        if 0 <= nr < 3 and 0 <= nc < 3:
            ni = nr * 3 + nc
            new_state = list(state)
            new_state[i], new_state[ni] = new_state[ni], new_state[i]
            moves.append(tuple(new_state))
    return moves

# A* search
def astar(start, goal):
    open_list = []
    heapq.heappush(open_list, (misplaced(start, goal), 0, start, None))
    closed = {}

    while open_list:
        f, g, state, parent = heapq.heappop(open_list)
        if state in closed:
            continue
        closed[state] = parent

        if state == goal:
            return reconstruct_path(closed, state)

        for nb in neighbors(state):
            if nb not in closed:
                heapq.heappush(open_list, (g+1+misplaced(nb, goal), g+1, nb, state))
    return None

# Reconstruct path
def reconstruct_path(closed, state):
    path = []
    while state:
        path.append(state)
        state = closed[state]
    return path[::-1]

# Example usage
if __name__ == "__main__":
    start = (1, 2, 3,
             4, 0, 6,
             7, 5, 8)
    goal = (1, 2, 3,
            4, 5, 6,
            7, 8, 0)

    path = astar(start, goal)
    if path:
        print(f"Solution found in {len(path)-1} moves:")
        for step in path:
            for i in range(0, 9, 3):
                print(step[i:i+3])
            print()
    else:
        print("No solution.")
```

```python
import heapq

# Manhattan distance heuristic
def manhattan(state, goal):
    dist = 0
    for i, tile in enumerate(state):
        if tile != 0:  # skip blank
            gi = goal.index(tile)
            dist += abs(i // 3 - gi // 3) + abs(i % 3 - gi % 3)
    return dist

# Generate neighbors by sliding blank
def neighbors(state):
    moves = []
    i = state.index(0)  # blank position
    r, c = divmod(i, 3)
    directions = [(-1,0),(1,0),(0,-1),(0,1)]
    for dr, dc in directions:
        nr, nc = r + dr, c + dc
        if 0 <= nr < 3 and 0 <= nc < 3:
            ni = nr * 3 + nc
            new_state = list(state)
            new_state[i], new_state[ni] = new_state[ni], new_state[i]
            moves.append(tuple(new_state))
    return moves

# A* search
def astar(start, goal):
    open_list = []
    heapq.heappush(open_list, (manhattan(start, goal), 0, start, None))
    closed = {}

    while open_list:
        f, g, state, parent = heapq.heappop(open_list)
        if state in closed:
            continue
        closed[state] = parent

        if state == goal:
            return reconstruct_path(closed, state)

        for nb in neighbors(state):
            if nb not in closed:
                heapq.heappush(open_list, (g+1+manhattan(nb, goal), g+1, nb, state))
    return None

# Reconstruct path from closed set
def reconstruct_path(closed, state):
    path = []
    while state:
        path.append(state)
        state = closed[state]
    return path[::-1]

# Example usage
if __name__ == "__main__":
    start = (1, 2, 3,
             4, 0, 6,
             7, 5, 8)
    goal = (1, 2, 3,
            4, 5, 6,
            7, 8, 0)

    path = astar(start, goal)
    if path:
        print(f"Solution found in {len(path)-1} moves:")
        for step in path:
            for i in range(0, 9, 3):
                print(step[i:i+3])
            print()
    else:
        print("No solution.")

```
