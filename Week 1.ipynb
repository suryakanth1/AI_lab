```python
import random
import time

class VacuumEnvironment:
    def __init__(self):
        self.location_status = {
            'A': random.choice(['Clean', 'Dirty']),
            'B': random.choice(['Clean', 'Dirty'])
        }
        self.vacuum_location = random.choice(['A', 'B'])

    def display_status(self):
        print(f"Current Environment Status:")
        print(f"   Room A: {self.location_status['A']}")
        print(f"   Room B: {self.location_status['B']}")
        print(f"    Vacuum Position: Room {self.vacuum_location}")
        print("-" * 40)

class ReflexVacuumAgent:
    def __init__(self):
        self.performance_score = 0

    def act(self, environment):
        current_room = environment.vacuum_location
        current_status = environment.location_status[current_room]

        print(f"🔎 Sensing: Vacuum is in Room {current_room}. Status: {current_status}")
        time.sleep(1)
        if current_status == 'Dirty':
            print(f" Action: Room {current_room} is Dirty! Sucking dirt...")
            environment.location_status[current_room] = 'Clean'
            self.performance_score += 10
            print(f" Room {current_room} is now Clean.")
        elif current_room == 'A':
            print("Action: Room A is Clean. Moving Right to Room B...")
            environment.vacuum_location = 'B'
            self.performance_score -= 1

        elif current_room == 'B':
            print("⬅ Action: Room B is Clean. Moving Left to Room A...")
            environment.vacuum_location = 'A'
            self.performance_score -= 1

def run_simulation():

    env = VacuumEnvironment()
    agent = ReflexVacuumAgent()

    print("=" * 40)
    print(" REFLEX VACUUM CLEANER AGENT STARTING ")
    print("=" * 40)

    step = 1
    while env.location_status['A'] == 'Dirty' or env.location_status['B'] == 'Dirty':
        print(f"\n--- STEP {step} ---")
        env.display_status()
        agent.act(env)
        step += 1

    print("\n" + "=" * 40)
    print(" SUCCESS: All rooms are clean!")
    env.display_status()
    print(f" Final Agent Performance Score: {agent.performance_score}")
    print("=" * 40)

if __name__ == "__main__":
    run_simulation()
```

```python
import random
def print_board(board):
    print("\n")
    print(f" {board[0]} | {board[1]} | {board[2]} ")
    print("-----------")
    print(f" {board[3]} | {board[4]} | {board[5]} ")
    print("-----------")
    print(f" {board[6]} | {board[7]} | {board[8]} ")
    print("\n")
def check_win(board, player):
    win_conditions = [
        [0, 1, 2], [3, 4, 5], [6, 7, 8],
        [0, 3, 6], [1, 4, 7], [2, 5, 8],
        [0, 4, 8], [2, 4, 6]
    ]
    for condition in win_conditions:
        if board[condition[0]] == board[condition[1]] == board[condition[2]] == player:
            return True
    return False
def check_tie(board):
    return " " not in board
def get_player_move(board):
    while True:
        try:
            move = int(input("Enter your move (1-9): ")) - 1
            if move in range(0, 9) and board[move] == " ":
                return move
            else:
                print("Invalid move. That spot is either taken or out of range.")
        except ValueError:
            print("Please enter a valid number between 1 and 9.")
def get_computer_move(board):
    for i in range(9):
        if board[i] == " ":
            board_copy = board.copy()
            board_copy[i] = "O"
            if check_win(board_copy, "O"):
                return i
    for i in range(9):
        if board[i] == " ":
            board_copy = board.copy()
            board_copy[i] = "X"
            if check_win(board_copy, "X"):
                return i
    if board[4] == " ":
        return 4
    empty_spots = [i for i, spot in enumerate(board) if spot == " "]
    return random.choice(empty_spots)
def play_game():
    board = [" "] * 9

    print("Welcome to Tic-Tac-Toe!")
    print("Positions match your keyboard grid layout from top-left (1) to bottom-right (9):")
    print(" 1 | 2 | 3 \n-----------\n 4 | 5 | 6 \n-----------\n 7 | 8 | 9 ")
    current_turn = random.choice(["Player", "Computer"])
    print(f"\n{current_turn} goes first!")
    print_board(board)

    while True:
        if current_turn == "Player":
            move = get_player_move(board)
            board[move] = "X"
            print_board(board)

            if check_win(board, "X"):
                print(" Congratulations! You beat the computer!")
                break

            if check_tie(board):
                print(" It's a tie game!")
                break

            current_turn = "Computer"

        else:
            print("Computer is thinking...")
            move = get_computer_move(board)
            board[move] = "O"
            print_board(board)

            if check_win(board, "O"):
                print(" Game over! The computer wins.")
                break

            if check_tie(board):
                print(" It's a tie game!")
                break

            current_turn = "Player"

if __name__ == "__main__":
    play_game()
```
