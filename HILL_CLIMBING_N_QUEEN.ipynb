```python
import random

def calculate_conflicts(state):
    """Calculates total row and diagonal conflicts on the board."""
    n = len(state)
    conflicts = 0
    for i in range(n):
        for j in range(i + 1, n):
            if state[i] == state[j]:
                conflicts += 1
            if abs(state[i] - state[j]) == j - i:
                conflicts += 1
    return conflicts

def get_best_neighbor(state):
    """Finds the neighbor with the minimum number of conflicts."""
    n = len(state)
    current_conflicts = calculate_conflicts(state)
    best_state = list(state)
    best_conflicts = current_conflicts

    for col in range(n):
        orig_row = state[col]
        for row in range(n):
            if row == orig_row:
                continue
            neighbor = list(state)
            neighbor[col] = row
            neighbor_conflicts = calculate_conflicts(neighbor)

            if neighbor_conflicts < best_conflicts:
                best_conflicts = neighbor_conflicts
                best_state = neighbor

    return best_state, best_conflicts

def print_board(state):
    """Renders the 1D state array as a 2D text matrix."""
    n = len(state)
    for row in range(n):
        row_str = ""
        for col in range(n):
            if state[col] == row:
                row_str += " Q "
            else:
                row_str += " . "
        print(row_str)

def run_hill_climbing_4_queens():
    # Targeted initial state configuration
    current_state = [0, 0, 0, 2]
    step = 1

    while True:
        current_conflicts = calculate_conflicts(current_state)
        print(f"=== STEP {step} ===")
        print(f"CURRENT STATE        : {current_state} (Conflicts: {current_conflicts})")

        neighbor_state, neighbor_conflicts = get_best_neighbor(current_state)

        # If no neighborhood layout reduces conflicts, we have hit a peak
        if neighbor_conflicts >= current_conflicts:
            print("CHOOSEN NEXT STATE   : None (Stuck or optimal peak reached)")
            print("=" * 45)
            break

        print(f"CHOOSEN NEXT STATE   : {neighbor_state} (Conflicts: {neighbor_conflicts})")
        print("=" * 45)

        current_state = neighbor_state
        step += 1

    print(f"\nOPTIMAL GLOBAL STATE FOUND")
    print(f"Final State Vector: {current_state}")
    print(f"Final Conflicts   : {calculate_conflicts(current_state)}")
    print("\nOptimal 2D Visual Board Layout:")
    print_board(current_state)

if __name__ == "__main__":
    run_hill_climbing_4_queens()
```
