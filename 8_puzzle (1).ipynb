```python

GOAL_STATE = (1, 2, 3,
              4, 5, 6,
              7, 8, 0)

def get_moves(state):
    moves = []
    zero_idx = state.index(0)
    row, col = zero_idx // 3, zero_idx % 3


    directions = [(-1, 0), (1, 0), (0, -1), (0, 1)] # Up, Down, Left, Right

    for dr, dc in directions:
        new_row, new_col = row + dr, col + dc
        if 0 <= new_row < 3 and 0 <= new_col < 3:
            swap_idx = new_row * 3 + new_col
            new_state = list(state)
            new_state[zero_idx], new_state[swap_idx] = new_state[swap_idx], new_state[zero_idx]
            moves.append(tuple(new_state))
    return moves

def dfs_8_puzzle(start_state):

    stack = [(start_state, [start_state])]
    visited = set()

    nodes_explored = 0

    while stack:
        current_state, path = stack.pop()(pops the deepest node)
        nodes_explored += 1

        if current_state == GOAL_STATE:
            print(f"Nodes explored: {nodes_explored}")
            return path

        if current_state not in visited:
            visited.add(current_state)

            # Explore neighbors
            for next_state in get_moves(current_state):
                if next_state not in visited:
                    stack.append((next_state, path + [next_state]))

    return None

def print_board(state):
    for i in range(0, 9, 3):
        print(f" {state[i] if state[i]!=0 else '.'} {state[i+1] if state[i+1]!=0 else '.'} {state[i+2] if state[i+2]!=0 else '.'} ")
    print()

# --- RUNNING THE SOLVER ---
# A simple configuration requiring a few moves
initial_board = (1, 2, 3,
                 4, 0, 5,
                 7, 8, 6)

print("Initial State:")
print_board(initial_board)

print("Searching using DFS...")
solution_path = dfs_8_puzzle(initial_board)

if solution_path:
    print(f"\nSolution Found in {len(solution_path) - 1} steps!")
    for step_num, state in enumerate(solution_path):
        print(f"Step {step_num}:")
        print_board(state)
else:
    print("No solution found.")
```

```python

GOAL_STATE = (1, 2, 3,
              4, 5, 6,
              7, 8, 0)

def get_moves(state):
    moves = []
    zero_idx = state.index(0)
    row, col = zero_idx // 3, zero_idx % 3

    directions = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in directions:
        new_row, new_col = row + dr, col + dc
        if 0 <= new_row < 3 and 0 <= new_col < 3:
            swap_idx = new_row * 3 + new_col

            new_state = list(state)
            new_state[zero_idx], new_state[swap_idx] = new_state[swap_idx], new_state[zero_idx]
            moves.append(tuple(new_state))
    return moves

def dls(current_state, limit, path):
    if current_state == GOAL_STATE:
        return path + [current_state]
    if limit <= 0:
        return None

    for next_state in get_moves(current_state):
        if next_state not in path: # Avoid cyclic loops in the current path
            result = dls(next_state, limit - 1, path + [current_state])
            if result is not None:
                return result
    return None


def iddfs_8_puzzle(start_state):
    depth = 0
    while True:
        print(f"Searching at depth limit: {depth}...")
        result = dls(start_state, depth, [])
        if result is not None:
            return result
        depth += 1

        if depth > 31:
            return None

def print_board(state):
    for i in range(0, 9, 3):
        print(f" {state[i] if state[i]!=0 else '.'} {state[i+1] if state[i+1]!=0 else '.'} {state[i+2] if state[i+2]!=0 else '.'} ")
    print()


initial_board = (1, 2, 3,
                 4, 0, 5,
                 7, 8, 6)

print("Initial State:")
print_board(initial_board)

solution_path = iddfs_8_puzzle(initial_board)

if solution_path:
    print(f"\nSolution Found in {len(solution_path) - 1} steps!")
    for step_num, state in enumerate(solution_path):
        print(f"Step {step_num}:")
        print_board(state)
else:
    print("No solution found or state is unsolvable.")
```
